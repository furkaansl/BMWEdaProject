```python
model_fiyat=df.groupby('model')['price'].mean().sort_values(ascending=False)
```

```python
model_fiyat.head(10)
```

```python
from numpy._core.defchararray import count
from numpy import mean

model_analiz = df.groupby('model')['price'].agg(['mean','count'])
```

```python
model_analiz = model_analiz.sort_values(by='mean', ascending=False)
```

```python
model_analiz.head(10)
```

```python
model_analiz['mean'].head(10).plot(kind='bar')
plt.title("BMW Model Bazlı Ortalama Fiyatlar")
plt.show()
```

```python
model_analiz['mean'].head(10).sort_values(ascending=False).plot(kind='barh')
plt.title("BMW Model Ortalama Fiyatta Küçükten Büyüğe Fiyatlar")
plt.show()
```
